# Finite MC systematics sample size
## Calculate how large sample size we need for MC

### Sample sizes

```c++
std::vector<double> sampleStats;
for(double stats = 1.0; stats <= 10.5; stats += 1.0) {
    sampleStats.push_back(stats);
}
```

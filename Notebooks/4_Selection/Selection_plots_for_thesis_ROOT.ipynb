### Plot MC and AmpGen samples of $D^0$ model

```c++
std::pair<double, double> PairPower(std::pair<double, double> x, int exponent) {
    x.first = TMath::Power(x.first, exponent);
    x.second = TMath::Power(x.second, exponent);
    return x;
}
```

```c++
std::vector<std::string> Variables{"s01", "s03", "s12", "s23", "s012"};
constexpr double m_pi = 0.13957;
constexpr double m_K = 0.493677;
constexpr double m_D = 1.86484;
const std::map<std::string, std::pair<double, double> > Limits{
    {"s01", PairPower(std::make_pair(2*m_K, m_D - 2*m_pi), 2)},
    {"s03", PairPower(std::make_pair(m_K + m_pi, m_D - m_K - m_pi), 2)},
    {"s12", PairPower(std::make_pair(m_K + m_pi, m_D - m_K - m_pi), 2)},
    {"s23", PairPower(std::make_pair(2*m_pi, m_D - 2*m_K), 2)},
    {"s012", PairPower(std::make_pair(2*m_K + m_pi, m_D - m_pi), 2)}
};
const std::map<std::string, std::string> Labels{
    {"s01", "K^{+}K^{#minus}"},
    {"s03", "K^{+}#pi^{#minus}"},
    {"s12", "K^{#minus}#pi^{+}"},
    {"s23", "#pi^{+}#pi^{#minus}"},
    {"s012", "K^{+}K^{#minus}#pi^{+}"}
};
```

```c++
TChain AmpGenChain_D0("DalitzTree");
AmpGenChain_D0.Add("/data/bes3/tat/KKpipi_StrongPhase_Analysis_cisi/ReweighterSamples/D02KKpipi_D0.root");
TChain BESIIIChain_D0("TruthTuple");
std::string Filename("/data/bes3/tat/KKpipi_StrongPhase_Analysis_4Bins_16fb/TruthTuples/BinnedTruthTuples/");
Filename += "KSpipi/KKpipi_vs_KSpipi_TruthTuple_Binned_Reweighted.root";
BESIIIChain_D0.Add(Filename.c_str());
for(const auto &Variable : Variables) {
    TH1D *h9a = new TH1D(("h9a_" + Variable).c_str(), "", 100,
                         Limits.at(Variable).first, Limits.at(Variable).second);
    AmpGenChain_D0.Draw((Variable + " >> h9a_" + Variable).c_str(),
                        "!(s23 > 0.477^2 && s23 < 0.507^2)", "GOFF");
    TH1D *h9b = new TH1D(("h9b_" + Variable).c_str(), "", 100,
                         Limits.at(Variable).first, Limits.at(Variable).second);
    BESIIIChain_D0.Draw((Variable + " >> h9b_" + Variable).c_str(), "", "GOFF");
    TCanvas *c9 = new TCanvas(("c9_" + Variable).c_str(), "", 1200, 900);
    h9a->Scale(10.0/h9a->Integral());
    h9b->Scale(10.0/h9b->Integral());
    h9a->SetTitle((";#it{m^{2}(" + Labels.at(Variable) + ")} (GeV^{2}/c^{4});Normalised entries").c_str());
    h9a->SetLineWidth(3);
    h9b->SetLineWidth(3);
    h9a->SetLineColor(kBlue);
    h9b->SetLineColor(kBlack);
    h9a->Draw("HIST");
    h9b->Draw("HIST SAME");
    if(Variable == "s01") {
        TLegend *Legend = new TLegend(0.5, 0.80, 0.9, 0.94);
        Legend->AddEntry(h9a, "AmpGen flavour D^{0} model", "L");
        Legend->AddEntry(h9b, "BESIII simulation", "L");
        Legend->Draw("SAME");
    }
    c9->Draw();
    c9->SaveAs(("ThesisSelectionPlot_D0_" + Variable + "_comparison.pdf").c_str());
}
```

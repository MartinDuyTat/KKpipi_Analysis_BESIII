# $\Delta E$ cuts
## Make tables of $\Delta E$ cuts

### Function for parsing and options file with the $\Delta E$ cuts

```c++
std::map<std::string, double> GetDeltaECuts(const std::string &Filename) {
    std::string Line;
    std::map<std::string, double> Cuts;
    std::ifstream File(Filename);
    while(std::getline(File, Line)) {
        if(Line.empty() || Line[0] == '*') {
            continue;
        }
        std::string Name;
        double Value;
        std::stringstream ss(Line);
        ss >> Name >> Value;
        Cuts.insert({Name, Value});
    }
    File.close();
    return Cuts;
}
```

```c++
std::string Dir("/data/bes3/tat/KKpipi_StrongPhase_Analysis_4Bins/CommonInputs/DataSets/");
auto DataCuts = GetDeltaECuts(Dir + "DeltaECuts_Data.txt");
auto MCCuts = GetDeltaECuts(Dir + "DeltaECuts_MC.txt");
```

### LaTeX names of all tag modes

```c++
std::vector<std::string> TagModes{"KKpipi",
                                  "KSpipi",
                                  "Kpi",
                                  "Kpipi0",
                                  "Kpipipi",
                                  "KK",
                                  "pipi",
                                  "KSpi0",
                                  "KSpi0pi0",
                                  "KSeta",
                                  "KSetaPrimepipieta",
                                  "KSetaPrimerhogamma",
                                  "pipipi0",
                                  "KSpipipi0"};
std::map<std::string, std::string> TagNames{{"KKpipi", "$\\kaonp\\kaonm\\pip\\pim$"},
                                            {"KSpipi", "$\\kshort\\pip\\pim$"},
                                            {"Kpi", "$\\kaonm\\pip$"},
                                            {"Kpipi0", "$\\kaonm\\pip\\piz$"},
                                            {"Kpipipi", "$\\kaonm\\pip\\pim\\pip$"},
                                            {"KK", "$\\kaonp\\kaonm$"},
                                            {"pipi", "$\\pip\\pim$"},
                                            {"KSpi0", "$\\kshort\\piz$"},
                                            {"KSpi0pi0", "$\\kshort\\piz\\piz$"},
                                            {"KSeta", "$\\kshort\\eta$"},
                                            {"KSetaPrimepipieta", "$\\kshort\\eta^\\prime_{\\pi\\pi\\eta}$"},
                                            {"KSetaPrimerhogamma", "$\\kshort\\eta^\\prime_{\\rho\\gamma}$"},
                                            {"pipipi0", "$\\pip\\pim\\piz$"},
                                            {"KSpipipi0", "$\\kshort\\pip\\pim\\piz$"}};
```

### Print tables

```c++
int BlankSpaces = 8;
std::string Blank = BlankSpaces == 0 ? "" : std::string(BlankSpaces, ' ');
int Width = 34;
std::cout << Blank << "\\toprule" << "\n";
std::cout << Blank;
std::cout << std::left << std::setw(Width) << "Tag mode" << " & "
          << std::left << std::setw(Width) << "$\\Delta E$ cut in data ($\\mev$)" << " & "
          << std::left << std::setw(Width) << "$\\Delta E$ cut in MC ($\\mev$)" << " \\\\" << "\n";
std::cout << Blank << "\\midrule" << "\n";
for(const auto &Tag : TagModes) {
    std::cout << Blank;
    std::cout << std::left << std::setw(Width);
    std::cout << TagNames[Tag] << " & ";
    std::cout << std::left << std::setw(Width);
    std::stringstream ssData;
    ssData << "$[" << std::fixed << std::setprecision(1);
    ssData << DataCuts[Tag + "_DeltaE_LowerCut"]*1000.0;
    ssData << "," << std::fixed << std::setprecision(1);
    ssData << DataCuts[Tag + "_DeltaE_UpperCut"]*1000.0;
    ssData << "]$";
    std::string DataLine;
    std::getline(ssData, DataLine);
    std::cout << DataLine << " & ";
    std::cout << std::left << std::setw(Width);
    std::stringstream ssMC;
    ssMC << "$[" << std::fixed << std::setprecision(1);
    ssMC << MCCuts[Tag + "_DeltaE_LowerCut"]*1000.0;
    ssMC << "," << std::fixed << std::setprecision(1);
    ssMC << MCCuts[Tag + "_DeltaE_UpperCut"]*1000.0;
    ssMC << "]$";
    std::string MCLine;
    std::getline(ssMC, MCLine);
    std::cout << MCLine << " \\\\" << "\n";
}
std::cout << Blank << "\\bottomrule" << "\n";
```

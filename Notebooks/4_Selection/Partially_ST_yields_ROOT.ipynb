### Number of $D^0\overline{D}^0$ pairs

````c++
uncertainties::udouble D0D0CrossSection(3615.0, TMath::Sqrt(10.0*10.0 + 38.0*38.0));
uncertainties::udouble LuminosityOldData(2931.8, TMath::Sqrt(0.2*0.2 + 13.8*13.8));
uncertainties::udouble Luminosity2022Data(4995.0, 19.0);
uncertainties::udouble Luminosity2023Data(8157.0, 31.0);
uncertainties::udouble Luminosity2024Data(4191.0, 16.0);
auto Luminosity = LuminosityOldData + Luminosity2022Data + Luminosity2023Data + Luminosity2024Data;
auto ND0D0 = D0D0CrossSection*Luminosity;
std::cout << "Luminosity = " << Luminosity << "\n";
std::cout << "N_D0D0 = " << ND0D0 << "\n";
````

````c++
std::map<std::string, std::string> Modes{
    {"KLpi0", "ModelWeight_CPOdd"},
    {"KLpipi", ""},
    {"KeNu", ""}
};
````

````c++
for(const auto &Mode : Modes) {
    auto DT_Eff = GetInclusiveDTEff(Mode.first, Mode.second);
    uncertainties::udouble DT_Eff_unc(DT_Eff.first, DT_Eff.second);
    auto KKpipi_ST_Eff = GetSTEfficiency("KKpipi");
    uncertainties::udouble KKpipi_ST_Eff_unc(KKpipi_ST_Eff.first, KKpipi_ST_Eff.second);
    auto ST_Eff = DT_Eff_unc/KKpipi_ST_Eff_unc;
    auto BranchingFraction = GetBranchingFraction(Mode.first);
    uncertainties::udouble BranchingFraction_unc(BranchingFraction.first, BranchingFraction.second);
    auto STyield = 2*ND0D0*BranchingFraction_unc*ST_Eff;
    std::cout << Mode.first << "_SingleTagEfficiency " << uncertainties::nom(ST_Eff) << "\n";
    std::cout << Mode.first << "_SingleTagEfficiency_err " << uncertainties::sdev(ST_Eff) << "\n";
    std::cout << Mode.first << "_SingleTag_Yield ";
    std::streamsize ss = std::cout.precision();
    std::cout << std::fixed << std::setprecision(1) << uncertainties::nom(STyield) << "\n";
    std::cout << Mode.first << "_SingleTag_Yield_err " << uncertainties::sdev(STyield) << "\n\n";
    std::cout << std::defaultfloat << std::setprecision(ss);
}
````

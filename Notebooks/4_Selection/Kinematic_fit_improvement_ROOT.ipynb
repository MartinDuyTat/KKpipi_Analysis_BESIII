# Kinematic fit improvement
## Demonstrate how effective the kinematic fit is

### Turn off statistics box

```c++
gStyle->SetOptStat(0);
```

### Add BESIII style

```c++
gInterpreter->AddIncludePath("/data/lhcb/users/tat/KKpipi_Analysis_BESIII/include");
```

```c++
gROOT->ProcessLine(".L /data/lhcb/users/tat/KKpipi_Analysis_BESIII/src/Bes3plotstyle.cpp");
```

```c++
SetStyle();
SetPrelimStyle();
```

### Load MC sample

```c++
TChain Chain("KSpipiDoubleTag");
std::string Path("/data/bes3/tat/KKpipi_StrongPhase_Analysis_4Bins/Selection/SignalMC/DoubleTag/");
Path += "KSpipi/KKpipi_vs_KSpipi_Binned_SignalMC.root";
Chain.Add(Path.c_str());
```

### Histogram of $K_+$ momentum

```c++
TH1D KPlusKalmanFit_hist("KPlusKalmanFit_hist", "", 100, -0.03, 0.03);
KPlusKalmanFit_hist.SetLineColor(kBlack);
KPlusKalmanFit_hist.SetLineWidth(3);
KPlusKalmanFit_hist.GetXaxis()->SetNdivisions(-404);
KPlusKalmanFit_hist.SetTitle(";p_{x}^{reco}-p_{x}^{true} (GeV/c);Entries");
KPlusKalmanFit_hist.GetYaxis()->SetMaxDigits(3);
Chain.Draw("SignalKPluspxKalmanFit - True_Px[2] >> KPlusKalmanFit_hist", "ParticleIDs[2] == 321", "GOFF");
TH1D KPlus_hist("KPlus_hist", "", 100, -0.03, 0.03);
KPlus_hist.SetLineColor(kBlue);
KPlus_hist.SetLineWidth(3);
Chain.Draw("SignalKPluspx - True_Px[2] >> KPlus_hist", "ParticleIDs[2] == 321", "GOFF");
TCanvas c1("c1", "", 1200, 900);
TLegend Legend_KPlus(0.63, 0.7, 0.94, 0.92);
KPlusKalmanFit_hist.Draw("HIST");
KPlus_hist.Draw("HIST SAME");
Legend_KPlus.AddEntry(&KPlus_hist, "Before kinematic fit", "L");
Legend_KPlus.AddEntry(&KPlusKalmanFit_hist, "After kinematic fit", "L");
Legend_KPlus.Draw();
TLatex Text;
Text.SetTextFont(42);
Text.SetTextSize(0.09);
Text.SetTextColor(kBlack);
Text.SetNDC(true);
Text.SetText(0.2, 0.8, "K^{+}");
Text.Draw("SAME");
c1.SetTopMargin(0.07);
c1.Draw();
c1.SaveAs("KPlus_kinematic_fit_comparison.pdf");
```

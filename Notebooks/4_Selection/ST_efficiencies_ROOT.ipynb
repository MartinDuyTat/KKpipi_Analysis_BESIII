### List of tag modes

````c++
const std::vector<std::string> TagModes{
    "KKpipi",
    "Kpi",
    "Kpipi0",
    "Kpipipi",
    "KK",
    "KSpi0",
    "pipipi0",
    "pipi",
    "KSeta",
    "KSpi0pi0",
    "KSpipipi0",
    "KSetaPrimepipieta",
    "KSetaPrimerhogamma",
    "KSpipi"
};
````

### Make table of tag efficiencies

````c++
std::map<std::string, double> SingleTagEfficiencies;
int BlankSpaces = 8;
std::string Blank = BlankSpaces == 0 ? "" : std::string(BlankSpaces, ' ');
int Width = 35;
std::cout << Blank << "\\toprule" << "\n";
std::cout << Blank;
std::cout << std::left << std::setw(Width) << "Tag mode" << " & "
          << std::left << std::setw(Width) << "Single tag efficiency ($10^{-2}$)" << " \\\\" << "\n";
std::cout << Blank << "\\midrule" << "\n";
for(const auto &Tag : TagModes) {
    auto Efficiency = GetSTEfficiency(Tag);
    SingleTagEfficiencies.insert({Tag, Efficiency.first});
    SingleTagEfficiencies.insert({Tag + "_err", Efficiency.second});
    PrintLaTeXLine(GetTagName(Tag), Efficiency.first*100, Efficiency.second*100, 0, BlankSpaces, Width);
}
std::cout << Blank << "\\bottomrule" << "\n";
````

### Write efficiencies to file

````c++
std::ofstream ST_Eff_File("SingleTagEfficiencies.txt");
for(const auto &Tag : TagModes) {
    ST_Eff_File << std::left << std::setw(45);
    ST_Eff_File << Tag + "_SingleTagEfficiency     " << SingleTagEfficiencies[Tag] << "\n";
    ST_Eff_File << std::left << std::setw(45);
    ST_Eff_File << Tag + "_SingleTagEfficiency_err " << SingleTagEfficiencies[Tag + "_err"] << "\n\n";
}
ST_Eff_File.close();
````

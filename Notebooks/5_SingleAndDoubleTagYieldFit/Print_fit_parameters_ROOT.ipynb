### Function for parsing fit parameters

```c++
std::map<std::string, double> ParseFitParameters(const std::string &Tag,
                                                 const std::string &ST_or_DT) {
    std::string Filename("/data/bes3/tat/KKpipi_StrongPhase_Analysis_4Bins_16fb/");
    if(ST_or_DT == "ST") {
        Filename += "SingleTagFit/" + Tag + "/" + Tag + "_MBC_FitResults.txt";
    } else {
        Filename += "DoubleTagFit/" + Tag + "/KKpipi_vs_" + Tag + "_SignalYield.txt";
    }
    std::map<std::string, double> FitParameters;
    std::ifstream File(Filename);
    std::string Line;
    while(std::getline(File, Line)) {
        if(Line.empty() || Line[0] == '*') {
            continue;
        }
        std::string Name;
        double Value;
        std::stringstream ss(Line);
        ss >> Name >> Value;
        FitParameters.insert({Name, Value});
    }
    return FitParameters;
}
```

```c++
ParseFitParameters("KKpipi", "ST")
```

### List of single tags

```c++
std::vector<std::string> STags{"KKpipi", "Kpi", "Kpipi0", "Kpipipi",
                               "KK", "pipi", "pipipi0", "KSpi0pi0",
                               "KSpi0", "KSeta", "KSetaPrimepipieta", "KSetaPrimerhogamma", "KSpipipi0",
                               "KSpipi"};
```

### Print table of single tag yield fit parameters

```c++
int BlankSpaces = 8;
std::string Blank = BlankSpaces == 0 ? "" : std::string(BlankSpaces, ' ');
int Width = 25;
for(const auto &Tag : STags) {
    const auto FitParameters = ParseFitParameters(Tag, "ST");
    std::cout << Tag << "\n";
    // Signal yield
    std::cout << Blank;
    std::cout << std::left << std::setw(20);
    std::cout << "$N_{\\rm sig}$" << " & ";
    std::stringstream ss;
    ss << "$" << std::fixed << std::setprecision(0) << FitParameters.at(Tag + "_Yield") << " \\pm ";
    ss << std::fixed << std::setprecision(0) << FitParameters.at(Tag + "_Yield_err") << "$" << " \\\\";
    std::cout << ss.str() << "\n";
    ss.clear();
    ss.str("");
    // Background yield
    std::cout << Blank;
    std::cout << std::left << std::setw(20);
    std::cout << "$N_{\\rm bkg}$" << " & ";
    ss << "$" << std::fixed << std::setprecision(0);
    ss << FitParameters.at(Tag + "_CombinatorialYield") << " \\pm ";
    ss << std::fixed << std::setprecision(0);
    ss << FitParameters.at(Tag + "_CombinatorialYield_err") << "$" << " \\\\";
    std::cout << ss.str() << "\n";
    ss.clear();
    ss.str("");
    if(FitParameters.find(Tag + "_Mean1") == FitParameters.end()) {
        // Single Gaussian resolution function
        // Mean of Gaussian
        std::cout << Blank;
        std::cout << std::left << std::setw(20);
        std::cout << "$\\mu$ (\\mev)" << " & ";
        std::cout << PrintLaTeXNumber(FitParameters.at(Tag + "_Mean")*1000.0,
                                      FitParameters.at(Tag + "_Mean_err")*1000.0);
        std::cout << " \\\\" << "\n";
        // Width of Gaussian
        std::cout << Blank;
        std::cout << std::left << std::setw(20);
        std::cout << "$\\sigma$ (\\mev)" << " & ";
        std::cout << PrintLaTeXNumber(FitParameters.at(Tag + "_Sigma")*1000.0,
                                      FitParameters.at(Tag + "_Sigma_err")*1000.0);
        std::cout << " \\\\" << "\n";
    } else {
        // Double Gaussian resolution
        // Mean of first Gaussian
        std::cout << Blank;
        std::cout << std::left << std::setw(20);
        std::cout << "$\\mu_1$ (\\mev)" << " & ";
        std::cout << PrintLaTeXNumber(FitParameters.at(Tag + "_Mean1")*1000.0,
                                      FitParameters.at(Tag + "_Mean1_err")*1000.0);
        std::cout << " \\\\" << "\n";
        // Mean of second Gaussian
        std::cout << Blank;
        std::cout << std::left << std::setw(20);
        std::cout << "$\\mu_2$ (\\mev)" << " & ";
        std::cout << PrintLaTeXNumber(FitParameters.at(Tag + "_Mean2")*1000.0,
                                      FitParameters.at(Tag + "_Mean2_err")*1000.0);
        std::cout << " \\\\" << "\n";
        // Width of first Gaussian
        std::cout << Blank;
        std::cout << std::left << std::setw(20);
        std::cout << "$\\sigma_1$ (\\mev)" << " & ";
        std::cout << PrintLaTeXNumber(FitParameters.at(Tag + "_Sigma2")*1000.0,
                                      FitParameters.at(Tag + "_Sigma2_err")*1000.0);
        std::cout << " \\\\" << "\n";
        // Width of second Gaussian
        std::cout << Blank;
        std::cout << std::left << std::setw(20);
        std::cout << "$\\sigma_2$ (\\mev)" << " & ";
        std::cout << PrintLaTeXNumber(FitParameters.at(Tag + "_Sigma2")*1000.0,
                                      FitParameters.at(Tag + "_Sigma2_err")*1000.0);
        std::cout << " \\\\" << "\n";
        // Fraction of first Gaussian
        std::cout << Blank;
        std::cout << std::left << std::setw(20);
        std::cout << "$f$" << " & ";
        std::cout << PrintLaTeXNumber(FitParameters.at(Tag + "_frac"),
                                      FitParameters.at(Tag + "_frac_err"));
        std::cout << " \\\\" << "\n";
    }
    // Argus parameter
    std::cout << Blank;
    std::cout << std::left << std::setw(20);
    std::cout << "$c$" << " & ";
    std::cout << PrintLaTeXNumber(FitParameters.at(Tag + "_c"),
                                  FitParameters.at(Tag + "_c_err"));
    std::cout << " \\\\" << "\n";
}
```

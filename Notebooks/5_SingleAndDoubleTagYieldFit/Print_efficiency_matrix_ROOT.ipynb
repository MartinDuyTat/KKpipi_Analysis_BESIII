### Function for printing the efficiency matrix

```c++
void PrintEfficiencyMatrix(const TMatrixT<double> &EffMatrix,
                           const TMatrixT<double> &EffMatrix_err,
                           const std::string &WhatToPrint = "") {
    if(EffMatrix.GetNrows() != EffMatrix.GetNcols() ||
       EffMatrix.GetNrows() != EffMatrix_err.GetNrows() ||
       EffMatrix.GetNcols() != EffMatrix_err.GetNcols()) {
        std::cout << "Invalid efficiency matrices\n";
        return;
    }
    int N = EffMatrix.GetNrows();
    int BlankSpaces = 8;
    std::string Blank = BlankSpaces == 0 ? "" : std::string(BlankSpaces, ' ');
    int Width = 18;
    std::vector<int> BinNumbers;
    if(N == 8) {
        for(int Bin = -4; Bin <= 4; Bin++) {
            if(Bin == 0) {
                continue;
            }
            BinNumbers.push_back(Bin);
        }
    } else if(N == 4) {
        for(int Bin = 1; Bin <= 4; Bin++) {
            BinNumbers.push_back(Bin);
        }
    } else {
        return;
    }
    for(int i = 0; i < N; i++) {
        std::cout << Blank;
        std::stringstream ss;
        ss << "$";
        if(EffMatrix.GetNrows() != 4 && BinNumbers[i] > 0) {
            ss << "+";
        }
        ss << BinNumbers[i] << "$";
        std::cout << std::left << std::setw(4) << ss.str() << " & ";
        for(int j = 0; j < N; j++) {
            std::cout << std::left << std::setw(Width);
            std::stringstream ss2;
            ss2 << "$";
            if(WhatToPrint == "" || WhatToPrint == "Value") {
                ss2 << std::fixed << std::setprecision(2);
                ss2 << EffMatrix(i, j)*100.0;
            }
            if(WhatToPrint == "") {
                ss2 << " \\pm ";
            }
            if(WhatToPrint == "" || WhatToPrint == "Error") {
                ss2 << std::fixed << std::setprecision(2);
                ss2 << EffMatrix_err(i, j)*100.0;
            }
            ss2 << "$";
            std::cout << ss2.str();
            if(j == N - 1) {
                std::cout << " \\\\" << "\n";
            } else {
                std::cout << " & ";
            }
        }
    }
}
```

### Function for printing efficiency matrix with bin pairs

```c++
void PrintSCMBEfficiencyMatrix(const TMatrixT<double> &EffMatrix,
                               const TMatrixT<double> &EffMatrix_err,
                               const std::string &WhatToPrint = "") {
    if(EffMatrix.GetNrows() != EffMatrix.GetNcols() ||
       EffMatrix.GetNrows() != EffMatrix_err.GetNrows() ||
       EffMatrix.GetNcols() != EffMatrix_err.GetNcols()) {
        std::cout << "Invalid efficiency matrices\n";
        return;
    }
    std::vector<std::pair<int, int> > BinNumbers;
    for(int TagBin = 1; TagBin <= 8; TagBin++) {
        for(int SignalBin = -4; SignalBin <= 4; SignalBin++) {
            if(SignalBin == 0) {
                continue;
            }
            BinNumbers.push_back({SignalBin, TagBin});
        }
    }
    int N = EffMatrix.GetNrows();
    int BlankSpaces = 8;
    std::string Blank = BlankSpaces == 0 ? "" : std::string(BlankSpaces, ' ');
    int Width = 6;
    for(int i = 0; i < 64; i++) {
        int SignalBin = BinNumbers[i].first;
        int TagBin = BinNumbers[i].second;
        std::cout << Blank;
        std::stringstream ss;
        ss << "$(";
        if(SignalBin > 0) {
            ss << "+";
        }
        ss << SignalBin << "," << TagBin << ")$";
        std::cout << std::left << std::setw(4) << ss.str() << " & ";
        for(int j = 0; j < 64; j++) {
            std::cout << std::left << std::setw(Width);
            std::stringstream ss2;
            ss2 << "$";
            if(WhatToPrint == "" || WhatToPrint == "Value") {
                ss2 << std::fixed << std::setprecision(2);
                ss2 << EffMatrix(i, j)*100.0;
            }
            if(WhatToPrint == "") {
                ss2 << " \\pm ";
            }
            if(WhatToPrint == "" || WhatToPrint == "Error") {
                ss2 << std::fixed << std::setprecision(2);
                ss2 << EffMatrix_err(i, j)*100.0;
            }
            ss2 << "$";
            std::cout << ss2.str();
            if(j == 63) {
                std::cout << " \\\\" << "\n";
            } else {
                std::cout << " & ";
            }
        }
    }
}
```

### Function for printing flavour tag efficiency matrix

```c++
void PrintFlavourTagEffMatrix(const std::string &Tag) {
    std::string Filename("/data/bes3/tat/KKpipi_StrongPhase_Analysis_4Bins_20fb/CommonInputs/Efficiencies/");
    Filename += "EfficiencyMatrices/KKpipi_vs_" + Tag + "_EfficiencyMatrix.root";
    TFile File(Filename.c_str(), "READ");
    TMatrixT<double> *EffMatrix = nullptr;
    File.GetObject("EffMatrix", EffMatrix);
    TMatrixT<double> *EffMatrix_err = nullptr;
    File.GetObject("EffMatrix_err", EffMatrix_err);
    PrintEfficiencyMatrix(*EffMatrix, *EffMatrix_err);
}
```

### Function for printing CP tag efficiency matrix

```c++
void PrintCPTagEffMatrix(const std::string &Tag) {
    std::string Filename("/data/bes3/tat/KKpipi_StrongPhase_Analysis_4Bins_20fb/CommonInputs/Efficiencies/");
    Filename += "EfficiencyMatrices/KKpipi_vs_" + Tag + "_EfficiencyMatrix.root";
    TFile File(Filename.c_str(), "READ");
    TMatrixT<double> *EffMatrix_CPEven = nullptr;
    File.GetObject("EffMatrix_CPEven", EffMatrix_CPEven);
    TMatrixT<double> *EffMatrix_CPEven_err = nullptr;
    File.GetObject("EffMatrix_CPEven_err", EffMatrix_CPEven_err);
    TMatrixT<double> *EffMatrix_CPOdd = nullptr;
    File.GetObject("EffMatrix_CPOdd", EffMatrix_CPOdd);
    TMatrixT<double> *EffMatrix_CPOdd_err = nullptr;
    File.GetObject("EffMatrix_CPOdd_err", EffMatrix_CPOdd_err);
    double FPlus = GetFPlus(Tag).first;
    auto EffMatrix = (1.0 - FPlus)*(*EffMatrix_CPEven) + FPlus*(*EffMatrix_CPOdd);
    TMatrixT<double> EffMatrix_err(EffMatrix.GetNrows(), EffMatrix.GetNcols());
    for(int i = 0; i < EffMatrix.GetNrows(); i++) {
        for(int j = 0; j < EffMatrix.GetNcols(); j++) {
            EffMatrix_err(i, j) = TMath::Sqrt(TMath::Power((1.0 - FPlus)*(*EffMatrix_CPEven_err)(i, j), 2)
                                            + TMath::Power(FPlus*(*EffMatrix_CPOdd_err)(i, j), 2));
        }
    }
    PrintEfficiencyMatrix(EffMatrix, EffMatrix_err);
}
```

### Function for printing SCMB efficiency matrix

```c++
void PrintSCMBTagEffMatrix(const std::string &Tag) {
    std::string Filename("/data/bes3/tat/KKpipi_StrongPhase_Analysis_4Bins_20fb/CommonInputs/Efficiencies/");
    Filename += "EfficiencyMatrices/KKpipi_vs_" + Tag + "_EfficiencyMatrix.root";
    TFile File(Filename.c_str(), "READ");
    TMatrixT<double> *EffMatrix = nullptr;
    File.GetObject("EffMatrix", EffMatrix);
    TMatrixT<double> *EffMatrix_err = nullptr;
    File.GetObject("EffMatrix_err", EffMatrix_err);
    std::cout << "Value:\n";
    PrintSCMBEfficiencyMatrix(*EffMatrix, *EffMatrix_err, "Value");
    std::cout << "Error:\n";
    PrintSCMBEfficiencyMatrix(*EffMatrix, *EffMatrix_err, "Error");
}
```

### Print bin numbers for LaTeX tables

```c++
int BlankSpaces = 8;
std::string Blank = BlankSpaces == 0 ? "" : std::string(BlankSpaces, ' ');
int Width = 6;
std::cout << Blank;
std::cout << std::left << std::setw(4) << "True bin $\\rightarrow$" << " & ";
for(int TagBin = 1; TagBin <= 8; TagBin++) {
    for(int SignalBin = -4; SignalBin <= 4; SignalBin++) {
        if(SignalBin == 0) {
            continue;
        }
        std::stringstream ss;
        ss << "$(";
        if(SignalBin > 0) {
            ss << "+";
        }
        ss << SignalBin << "," << TagBin << ")$";
        std::cout << std::left << std::setw(Width) << ss.str();
        if(SignalBin == 4 && TagBin == 8) {
            std::cout << " \\\\" << "\n";
        } else {
            std::cout << " & ";
        }
    }
}
std::cout << Blank;
std::cout << std::left << std::setw(4) << "Reconstructed bin $\\downarrow$" << " & ";
for(int TagBin = 1; TagBin <= 8; TagBin++) {
    for(int SignalBin = -4; SignalBin <= 4; SignalBin++) {
        if(SignalBin == 0) {
            continue;
        }
        std::cout << std::left << std::setw(Width) << " ";
        if(SignalBin == 4 && TagBin == 8) {
            std::cout << " \\\\" << "\n";
        } else {
            std::cout << " & ";
        }
    }
}
```

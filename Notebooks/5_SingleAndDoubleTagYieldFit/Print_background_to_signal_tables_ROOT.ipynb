# Print background to signal tables
## Make nice tables of background to signal ratios

### Load utility functions

```c++
gROOT->ProcessLine(".L ../UtilityFunctions.C");
```

### File with $K_SKK$ vs flavour tag backgrounds

```c++
auto ParsedParameters = ParseParameters("BackgroundToSignalRatios_Flavour_KSKK.txt");
```

### Tag modes with $K_SKK$ background

```c++
std::vector<std::string> Tags{"Kpi", "KeNu"};
```

### Print $K_SKK$ background table for flavour tags

```c++
int BlankSpaces = 8;
std::string Blank = BlankSpaces == 0 ? "" : std::string(BlankSpaces, ' ');
int Width = 14;
for(const auto &Tag : Tags) {
    std::cout << Blank;
    std::cout << std::left << std::setw(5) << GetTagName(Tag) << " & ";
    for(int Bin = -4; Bin <= 4; Bin++) {
        if(Bin == 0) {
            continue;
        }
        std::string Label(Tag + "_PeakingBackground0_DoubleTag_Flavour_KKpipi_vs_");
        Label += Tag + "_SignalBin" + (Bin > 0 ? "P" : "M") + std::to_string(TMath::Abs(Bin));
        Label += "_TagBin0_BackgroundToSignalRatio";
        double Value = ParsedParameters.at(Label);
        double Error = ParsedParameters.at(Label + "_err");
        std::cout << std::left << std::setw(Width) << PrintLaTeXNumber(Value*100, Error*100);
        if(Bin == 4) {
            std::cout << " \\\\" << "\n";
        } else {
            std::cout << " & ";
        }
    }
}
```

### File with peaking backgrounds on tag side for flavour tags

```c++
auto ParsedParameters = ParseParameters("BackgroundToSignalRatios_Flavour.txt");
```

### Tag modes with backgrounds on tag side

```c++
std::vector<std::string> Tags{"Kpipipi", "KeNu"};
std::map<std::string, std::vector<std::string> > Backgrounds{
    {"Kpipipi", {"$\\kshort\\kaonpm\\pimp$"}},
    {"KeNu", {"$\\kaonm\\piz e^{+}\\nu_e$", "$\\kaonm\\mu^{+}\\nu_\\mu$", "$\\kaonm\\pip\\piz$"}}
};
```

### Print flavour tag peaking backgrounds on the tag side

```c++
int BlankSpaces = 8;
std::string Blank = BlankSpaces == 0 ? "" : std::string(BlankSpaces, ' ');
int Width = 14;
for(const auto &Tag : Tags) {
    for(std::size_t i = 0; i < Backgrounds.at(Tag).size(); i++) {
        std::cout << Blank;
        std::cout << std::left << std::setw(5) << GetTagName(Tag) << " & ";
        std::cout << std::left << std::setw(5) << Backgrounds.at(Tag)[i] << " & ";
        for(int Bin = -4; Bin <= 4; Bin++) {
            if(Bin == 0) {
                continue;
            }
            std::string Label(Tag + "_PeakingBackground" + std::to_string(i + 1));
            Label += "_DoubleTag_Flavour_KKpipi_vs_" + Tag + "_SignalBin" + (Bin > 0 ? "P" : "M");
            Label += std::to_string(TMath::Abs(Bin)) + "_TagBin0_BackgroundToSignalRatio";
            double Value = ParsedParameters.at(Label);
            double Error = ParsedParameters.at(Label + "_err");
            std::cout << std::left << std::setw(Width) << PrintLaTeXNumber(Value*100, Error*100);
            if(Bin == 4) {
                std::cout << " \\\\" << "\n";
            } else {
                std::cout << " & ";
            }
        }
    }
}
```

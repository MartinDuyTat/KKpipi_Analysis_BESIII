````c++
#include<uncertainties/impl.hpp>
#include<uncertainties/ureal.hpp>
#include<uncertainties/io.hpp>
````

### Number of $D\bar{D}$ pairs

````c++
uncertainties::udouble N_DD(5.81, 0.06);
N_DD *= TMath::Power(10, 7);
````

### Function for multiplying branching fractions

````c++
template<class T, class... Ts>
T GetPredictedSTYield(T NumberDD, T Efficiency, T BF) {
    return 2*NumberDD*Efficiency*BF;
}
````

````c++
template<class T, class... Ts>
T GetPredictedSTYield(T NumberDD, T Efficiency, T FirstBF, Ts... BFs) {
    return FirstBF*GetPredictedSTYield(NumberDD, Efficiency, BFs...);
}
````

### Function for printing the yield

````c++
template<class T, class... Ts>
void PrintPredictedSTYield(T NumberDD, T Efficiency, Ts... BFs) {
    T STYield = GetPredictedSTYield(NumberDD, Efficiency, BFs...);
    std::cout << "$";
    std::cout << std::fixed << std::setprecision(0);
    std::cout << uncertainties::nom(STYield) << " \\pm ";
    std::cout << std::fixed << std::setprecision(0);
    std::cout << uncertainties::sdev(STYield);
    std::cout << "$\n";
    T Kpi_STYield = GetPredictedSTYield(N_DD,
                                           uncertainties::udouble(0.6869, 0.0007),
                                           uncertainties::udouble(0.03947, 0.00030));
    std::cout << STYield/Kpi_STYield << "\n";
}
````

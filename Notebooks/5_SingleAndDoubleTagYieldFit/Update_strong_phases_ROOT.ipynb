# Update strong phases
## Update the strong phases used to calculate quantum correlation factors

### Read in bias corrections

```c++
std::map<std::string, double> BiasCorrections;
std::ifstream BiasFile("../6_StrongPhaseFit/BiasCorrections.txt");
std::string Line;
while(std::getline(BiasFile, Line)) {
    std::stringstream ss(Line);
    std::string Name;
    double Value;
    ss >> Name >> Value;
    BiasCorrections.insert({Name, Value});
}
BiasFile.close();
```

### File with bias corrected strong phases

```c++
std::ofstream BiasCorrectedResults("BiasCorrectedResults.txt");
```

### File with fit results

```c++
std::string Filename("/data/bes3/tat/KKpipi_StrongPhase_Analysis_4Bins/cisiFit/");
Filename += "PreliminaryFitResults.root";
TFile FitResultsFile(Filename.c_str(), "READ");
std::vector<std::string> *VariableNames = nullptr;
FitResultsFile.GetObject("VariableNames", VariableNames);
std::vector<double> *FitValues = nullptr;
FitResultsFile.GetObject("FitValues", FitValues);
std::vector<double> *FitUncertainties = nullptr;
FitResultsFile.GetObject("FitUncertainties", FitUncertainties);
```

### Correct the fitted values

```c++
for(const auto &Parameter : *VariableNames) {
    if(Parameter[0] != 'c' && Parameter[0] != 's') {
        continue;
    }
    std::size_t Index = std::find(VariableNames->begin(), VariableNames->end(), Parameter)
                      - VariableNames->begin();
    double FittedValue = (*FitValues)[Index];
    double FittedUncertainty = (*FitUncertainties)[Index];
    double PullBias = BiasCorrections[Parameter + "_PullBias"];
    BiasCorrectedResults << Parameter << " " << FittedValue - PullBias*FittedUncertainty << "\n";
    BiasCorrectedResults << Parameter << "_err " << FittedUncertainty << "\n";
}
```

### Add model predicted values of $K_i$

```c++
BiasCorrectedResults << "K1 0.031164\n";
BiasCorrectedResults << "K2 0.101041\n";
BiasCorrectedResults << "K3 0.125911\n";
BiasCorrectedResults << "K4 0.0252633\n";
BiasCorrectedResults << "Kbar1 0.103099\n";
BiasCorrectedResults << "Kbar2 0.255377\n";
BiasCorrectedResults << "Kbar3 0.271768\n";
BiasCorrectedResults << "Kbar4 0.086376\n";
BiasCorrectedResults << "K1_err 0.0\n";
BiasCorrectedResults << "K2_err 0.0\n";
BiasCorrectedResults << "K3_err 0.0\n";
BiasCorrectedResults << "K4_err 0.0\n";
BiasCorrectedResults << "Kbar1_err 0.0\n";
BiasCorrectedResults << "Kbar2_err 0.0\n";
BiasCorrectedResults << "Kbar3_err 0.0\n";
BiasCorrectedResults << "Kbar4_err 0.0\n";
```

```c++
BiasCorrectedResults.close();
```

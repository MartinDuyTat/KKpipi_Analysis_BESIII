# Bias corrections
## Determine the bias corrections from toys

### Parameters we are interested in

````c++
std::vector<std::string> Parameters{"c1", "c2", "c3", "c4",
                                    "s1", "s2", "s3", "s4",
                                    "R_M4", "R_M3", "R_M2", "R_M1",
                                    "R_P1", "R_P2", "R_P3",
                                    "BF_KKpipi", "BF_KKpipi_KLpipi"
};
````

### Parameters with a simple bias correction

````c++
std::vector<std::string> SimpleParameters{"c1", "c2", "c3", "c4",
                                          "s1", "s2", "s3", "s4",
                                          "R_M4", "R_M3", "R_M2", "R_M1",
                                          "R_P1", "R_P2", "R_P3",
                                          "BF_KKpipi", "BF_KKpipi_KLpipi"
};
````

### Parameters with a complicated bias correction (asymmetric uncertainties)

````c++
std::vector<std::string> ComplicatedParameters{
};
````

### Vectors with toy values

````c++
std::map<std::string, std::vector<double>> ToyValues;
for(const auto &Parameter : Parameters) {
    ToyValues.insert({Parameter, std::vector<double>()});
    ToyValues[Parameter].reserve(1000);
}
````

### Load toy values

````c++
std::ifstream File("ToyValuesForBiasCorrections.txt");
std::string Line;
while(std::getline(File, Line)) {
    std::stringstream ss(Line);
    for(const auto &Parameter : Parameters) {
        double Value;
        ss >> Value;
        if(TMath::Abs(Value) < 5.0) {
            ToyValues[Parameter].push_back(Value);
        }
    }
}
````

### ```std::stringstream``` with bias corrections

````c++
std::stringstream BiasCorrections, Widths;
````

### Simple bias corections

````c++
for(const auto &Parameter : SimpleParameters) {
    double Bias = TMath::Mean(ToyValues[Parameter].begin(), ToyValues[Parameter].end());
    BiasCorrections << Parameter << "_PullBias " << Bias << "\n";
    double Width = TMath::RMS(ToyValues[Parameter].begin(), ToyValues[Parameter].end());
    Widths << Parameter << "_PullWidth " << Width << "\n";
}
````
